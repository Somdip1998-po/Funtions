1.  What is the difference between a function and a method in Python?
->**Function**

A function is a standalone block of code that performs a specific task.

It is defined using the def keyword and not tied to an object.

You can call a function directly by its name.

**Method**

A method is a function that is associated with an object—more specifically, with a class or instance of a class.

It is also defined using the def keyword, but inside a class.

You call a method using the dot (.) notation on an object.

It typically takes self as its first parameter (for instance methods).

6.  Explain the concept of generators in Python and how they are defined.
-> A generator is a function that uses the **yield** keyword instead of return to produce a sequence of values, one at a time, on demand.

When the generator function is called, it doesn’t run the function immediately but returns a generator object, which can be iterated over.

def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

7. What are the advantages of using generators over regular functions?
-> **Memory Efficiency**
Generators yield one item at a time and do not store the entire data in memory, unlike regular functions that might return large lists or object.
**Lazy Evaluation**
Generators produce values only when requested, which is ideal for large computations or infinite sequences.

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
->** map()**
Purpose: Transforms each item in an iterable by applying a function to each element.

Output: A new iterable where each item is the result of the function applied to the corresponding element in the original iterable.

Use Case: When you want to apply a transformation or computation to all elements of an iterable.

Syntax:
python
Copy
Edit
map(function, iterable)
Example:
python
Copy
Edit
nums = [1, 2, 3, 4]
squared = map(lambda x: x ** 2, nums)
print(list(squared))  # Output: [1, 4, 9, 16]
 **reduce()**
Purpose: Performs a cumulative reduction on an iterable (i.e., it applies a function to pairs of items and reduces the iterable to a single value).

Output: A single result that is the cumulative application of the function.

Use Case: When you need to "accumulate" a result (e.g., sum, product, or any other aggregation) across all elements.

Syntax:
python
Copy
Edit
from functools import reduce
reduce(function, iterable, [initial])
function: A function that takes two arguments and returns a single value.

iterable: The iterable to reduce.

initial (optional): An initial value to start the reduction (if not provided, it takes the first item from the iterable).

Example:
python
Copy
Edit
from functools import reduce
nums = [1, 2, 3, 4]
result = reduce(lambda x, y: x + y, nums)
print(result)  # Output: 10 (1 + 2 + 3 + 4)
 **filter()**

Purpose: Filters elements from an iterable by applying a function that returns True or False for each element. Only elements that evaluate to True are included in the result.

Output: A new iterable (filter object) containing only the items that meet the condition.

Use Case: When you want to filter out elements based on a condition.

Syntax:
python
Copy
Edit
filter(function, iterable)

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

```python
def sum_of_evens(numbers):
    # Using filter to select even numbers and then summing them
    even_numbers = filter(lambda x: x % 2 == 0, numbers)
    return sum(even_numbers)
# Example usage:
numbers = [1, 2, 3, 4, 5, 6, 7, 8]
result = sum_of_evens(numbers)
print(result)
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.


```python
def is_prime(num):
    if num <= 1:
        return False  # Numbers less than or equal to 1 are not prime
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False  # If num is divisible by any number between 2 and sqrt(num), it's not prime
    return True

# Checking prime numbers between 1 and 200
primes = [num for num in range(1, 201) if is_prime(num)]
print(primes)
```

5.  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, num_terms):
        self.num_terms = num_terms  # Total number of terms to generate
        self.current = 0  # To keep track of the current term
        self.prev = 0  # Previous Fibonacci number
        self.next = 1  # Next Fibonacci number

    def __iter__(self):
        return self  # The object itself is the iterator

    def __next__(self):
        if self.current < self.num_terms:
            if self.current == 0:
                self.current += 1
                return 0  # The first Fibonacci number is 0
            elif self.current == 1:
                self.current += 1
                return 1  # The second Fibonacci number is 1
            else:
                fib = self.prev + self.next
                self.prev = self.next
                self.next = fib
                self.current += 1
                return fib
        else:
            raise StopIteration  # Stop when the sequence has generated the specified number of terms

# Example usage:
fib = FibonacciIterator(10)
for num in fib:
    print(num)
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.


```python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i

# Example usage:
exponent = 5
for power in powers_of_two(exponent):
    print(power)
```

7. Implement a generator function that reads a file line by line and yields each line as a string.

```python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()  # `strip()` removes any trailing newline characters

# Example usage:
file_path = 'example.txt'  # Replace with the path to your file
for line in read_file_line_by_line(file_path):
    print(line)
```

9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

```python
# List of temperatures in Celsius
celsius_temps = [0, 20, 37, 100]

# Conversion function
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

# Use map to apply the conversion to each temperature
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

print("Celsius:", celsius_temps)
print("Fahrenheit:", fahrenheit_temps)
```

10. . Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
# Input string
input_str = "Hello World"

# Define vowels
vowels = "aeiouAEIOU"

# Use filter to remove vowels
no_vowels = ''.join(filter(lambda char: char not in vowels, input_str))

print("Original string:", input_str)
print("String without vowels:", no_vowels)
```
